```python
import gradio as gr
import moonshine_onnx
from moonshine_onnx import MoonshineOnnxModel
import librosa
import numpy as np

# 2. Load the Model
print("Loading Moonshine model...")
model = MoonshineOnnxModel(model_name="moonshine/tiny")

# 3. Define the Transcribe Function (Robust Logic)
def transcribe_audio(audio_path):
    # 'audio_path' comes from Gradio as a string (e.g., "/tmp/audio.wav")
    if audio_path is None:
        return "No audio found."
    
    try:
        # Load the file using librosa
        # sr=16000 forces it to 16k Hz (Required by Moonshine)
        # mono=True forces it to 1 channel (Required by Moonshine)
        audio_data, _ = librosa.load(audio_path, sr=16000, mono=True)
        
        # Run inference
        # We pass the clean float32 array directly to the model
        segments = model.transcribe(audio_data, rate=16000)
        
        # Combine segments into one text
        full_text = " ".join([seg for seg in segments])
        return full_text
        
    except Exception as e:
        return f"Error processing audio: {str(e)}"

# 4. Launch the Web Interface
ui = gr.Interface(
    fn=transcribe_audio,
    # type="filepath" handles the saving logic for us
    inputs=gr.Audio(sources=["microphone", "upload"], type="filepath"),
    outputs="text",
    title="Lightweight Voice-to-Text (Stable)",
    description="Record or Upload. This version automatically converts all audio to 16kHz mono to prevent errors."
)

ui.launch(share=True, debug=True)
```
